```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("Brain Tumor.csv")


df = df.drop(columns=['Image'])


X_brain = df.drop(columns=['Class'])
y_brain = df['Class']


scaler = StandardScaler()
X_brain = scaler.fit_transform(X_brain)


X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_brain, y_brain, test_size=0.2, random_state=42
)

print("Shape:", X_train_b.shape, X_test_b.shape)
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=200),
    "SVM Linear": SVC(kernel='linear'),
    "SVM RBF": SVC(kernel='rbf'),
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Naive Bayes": GaussianNB(),
    "Gradient Boosting": GradientBoostingClassifier()
}
```

```python
def evaluate(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    acc = accuracy_score(y_test, pred)
    pre = precision_score(y_test, pred, average='weighted')
    rec = recall_score(y_test, pred, average='weighted')
    f1 = f1_score(y_test, pred, average='weighted')
    return acc, pre, rec, f1
```

```python
results_iris = []

for name, model in models.items():
    acc, pre, rec, f1 = evaluate(model, X_train_i, X_test_i, y_train_i, y_test_i)
    results_iris.append([name, acc, pre, rec, f1])

df_iris_results = pd.DataFrame(results_iris, columns=["Model", "Accuracy", "Precision", "Recall", "F1 Score"])
print(df_iris_results)
```

```python
results_brain = []

for name, model in models.items():
    acc, pre, rec, f1 = evaluate(model, X_train_b, X_test_b, y_train_b, y_test_b)
    results_brain.append([name, acc, pre, rec, f1])

df_brain_results = pd.DataFrame(results_brain, columns=["Model", "Accuracy", "Precision", "Recall", "F1 Score"])
print(df_brain_results)
```
