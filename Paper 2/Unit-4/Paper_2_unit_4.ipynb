```python
# Install required libraries
!pip install -q transformers datasets accelerate evaluate

# Import necessary modules
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
import numpy as np
import evaluate

# Load the IMDb dataset
dataset = load_dataset("imdb")

# Load the BERT tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Tokenization function
def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=512)

# Tokenize datasets
tokenized_datasets = dataset.map(tokenize_function, batched=True)

# Use small subsets for fast training
small_train_dataset = tokenized_datasets["train"].shuffle(seed=42).select(range(1000))
small_eval_dataset = tokenized_datasets["test"].shuffle(seed=42).select(range(200))

# Load model
model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=2)

# Training arguments (fixed: use 'eval_strategy' instead of 'evaluation_strategy')
training_args = TrainingArguments(
    output_dir="bert_imdb_trainer",
    eval_strategy="epoch",           # Fixed: was 'evaluation_strategy'
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=1,
    weight_decay=0.01,
    logging_dir="logs",
    logging_steps=10,
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
)

# Load accuracy metric
metric = evaluate.load("accuracy")

# Compute metrics
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return metric.compute(predictions=predictions, references=labels)

# Initialize Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=small_train_dataset,
    eval_dataset=small_eval_dataset,
    compute_metrics=compute_metrics,
)

# Train
trainer.train()
```

For text summarization
Swap the models (bart-large-cnn, t5-small, pegasus-xsum) and change
parameters and use trainer API

```python
# Install required libraries
!pip install -q transformers datasets evaluate rouge_score accelerate torch

import torch
from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, Seq2SeqTrainingArguments, Seq2SeqTrainer, DataCollatorForSeq2Seq
import numpy as np
import evaluate

# === CONFIG: 3 MODELS ===
models_to_train = [
    {"name": "facebook/bart-large-cnn", "batch_size": 4,  "epochs": 1, "prefix": ""},
    {"name": "t5-small",                "batch_size": 8,  "epochs": 1, "prefix": "summarize: "},
    {"name": "google/pegasus-xsum",     "batch_size": 4,  "epochs": 1, "prefix": ""}
]

# Load CNN/DailyMail dataset
dataset = load_dataset("cnn_dailymail", "3.0.0")
train_dataset = dataset["train"].shuffle(seed=42).select(range(1000))
eval_dataset = dataset["validation"].shuffle(seed=42).select(range(200))

# Load ROUGE
rouge = evaluate.load("rouge")

def compute_metrics(eval_pred):
    preds, labels = eval_pred
    decoded_preds = tokenizer.batch_decode(preds, skip_special_tokens=True)
    labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)
    result = rouge.compute(predictions=decoded_preds, references=decoded_labels, use_stemmer=True)
    return {k: round(v, 4) for k, v in result.items()}

# === TRAINING LOOP ===
for idx, config in enumerate(models_to_train):
    print(f"\n{'='*70}")
    print(f"TRAINING MODEL {idx+1}/3: {config['name']}")
    print(f"{'='*70}\n")

    MODEL_NAME = config["name"]
    BATCH_SIZE = config["batch_size"]
    EPOCHS = config["epochs"]
    PREFIX = config["prefix"]

    # Load tokenizer & model
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

    # Preprocess function
    def preprocess(examples):
        inputs = [PREFIX + doc for doc in examples["article"]]
        model_inputs = tokenizer(inputs, max_length=512, truncation=True, padding="max_length")
        with tokenizer.as_target_tokenizer():
            labels = tokenizer(examples["highlights"], max_length=128, truncation=True, padding="max_length")
        model_inputs["labels"] = labels["input_ids"]
        return model_inputs

    # Tokenize
    tokenized_train = train_dataset.map(preprocess, batched=True, remove_columns=["article", "highlights", "id"])
    tokenized_eval = eval_dataset.map(preprocess, batched=True, remove_columns=["article", "highlights", "id"])

    data_collator = DataCollatorForSeq2Seq(tokenizer, model=model)

    # Training args
    training_args = Seq2SeqTrainingArguments(
        output_dir=f"./cnn_model_{idx}_{MODEL_NAME.split('/')[-1]}",
        eval_strategy="epoch",
        save_strategy="epoch",
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE,
        num_train_epochs=EPOCHS,
        weight_decay=0.01,
        predict_with_generate=True,
        fp16=torch.cuda.is_available(),
        logging_steps=10,
        load_best_model_at_end=True,
        metric_for_best_model="rouge1",
        report_to="none",
        disable_tqdm=True,
    )

    # Trainer
    trainer = Seq2SeqTrainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_train,
        eval_dataset=tokenized_eval,
        tokenizer=tokenizer,
        data_collator=data_collator,
        compute_metrics=compute_metrics,
    )

    # Train
    trainer.train()

    # Save
    save_dir = f"./trained_cnn_{MODEL_NAME.split('/')[-1]}"
    model.save_pretrained(save_dir)
    tokenizer.save_pretrained(save_dir)
    print(f"Model saved to: {save_dir}\n")
```

For Question Answering Use RoBERTa variants, GPT-3/4 fine-tuned on SQuAD
dataset

```python
!pip install -q transformers datasets accelerate sentencepiece

from datasets import load_dataset
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
from transformers import TrainingArguments, Trainer, DefaultDataCollator
import torch
import os

os.environ["WANDB_DISABLED"] = "true"

dataset = load_dataset("squad")

tokenizer = AutoTokenizer.from_pretrained("roberta-base")
model = AutoModelForQuestionAnswering.from_pretrained("roberta-base")

def preprocess(example):
    inputs = tokenizer(
        example["question"],
        example["context"],
        max_length=384,
        truncation=True,
        padding="max_length",
        return_offsets_mapping=True
    )
    offset = inputs["offset_mapping"]
    answers = example["answers"]["text"][0]
    start_char = example["answers"]["answer_start"][0]
    end_char = start_char + len(answers)

    start_pos = 0
    end_pos = 0
    for i, (s, e) in enumerate(offset):
        if s <= start_char < e:
            start_pos = i
        if s < end_char <= e:
            end_pos = i

    inputs["start_positions"] = start_pos
    inputs["end_positions"] = end_pos
    inputs.pop("offset_mapping")
    return inputs

tokenized = dataset.map(preprocess, batched=False)

collator = DefaultDataCollator()

args = TrainingArguments(
    output_dir="./roberta-qa",
    per_device_train_batch_size=8,
    num_train_epochs=1,
    logging_steps=100,
    report_to=[]
)

trainer = Trainer(
    model=model,
    args=args,
    train_dataset=tokenized["train"],
    eval_dataset=tokenized["validation"],
    data_collator=collator,
    tokenizer=tokenizer
)

trainer.train()

from transformers import pipeline
qa = pipeline("question-answering", model=model, tokenizer=tokenizer)
```

Build a simple rule-based chatbot for FAQs.

```python
def chatbot(message):
    msg = message.lower()

    # Greetings
    if "hello" in msg or "hi" in msg:
        return "Hello! How can I help you today?"

    # About company
    elif "what is your name" in msg:
        return "I'm a simple FAQ chatbot."

    elif "what do you do" in msg:
        return "I answer frequently asked questions."

    # Timings
    elif "working hours" in msg or "open" in msg:
        return "We are open from 9 AM to 6 PM, Monday to Saturday."

    # Pricing
    elif "price" in msg or "cost" in msg:
        return "Pricing depends on the service. Basic plans start at $10/month."

    # Support
    elif "support" in msg or "help" in msg:
        return "You can contact support at support@example.com."

    # Location
    elif "location" in msg or "where are you" in msg:
        return "We are located in Chennai, India."

    # Goodbye
    elif "bye" in msg or "goodbye" in msg:
        return "Goodbye! Have a great day!"

    # Default fallback
    else:
        return "I'm not sure about that. Could you ask something else?"


# Chat loop
print("Type 'exit' to stop.\n")
while True:
    user = input("You: ")
    if user.lower() == "exit":
        print("Chatbot: Goodbye!")
        break
    print("Chatbot:", chatbot(user))
```

Apply various prompting techniques and view the difference

```python
!pip install transformers accelerate sentencepiece -q

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

model = "google/flan-t5-large"  # fits on Colab T4 with 16GB
pipe = pipeline("text2text-generation", model=model, tokenizer=model, max_new_tokens=150)

def run(prompt):
    print("\n--- PROMPT ---")
    print(prompt)
    print("--- OUTPUT ---")
    print(pipe(prompt)[0]["generated_text"])
    print("\n")
```

```python


# 1. Zero-shot
run("Explain blockchain in simple terms.")

# 2. One-shot
run("""
Q: What is AI?
A: AI means machines that can think and learn like humans.

Q: What is blockchain?
A:
""")

# 3. Few-shot
run("""
Q: What is AI?
A: AI refers to machines performing tasks that require human intelligence.

Q: What is Machine Learning?
A: ML is a subset of AI that learns patterns from data.

Q: What is blockchain?
A:
""")

# 4. Chain-of-thought
run("Why do objects fall to the ground? Let's think step by step.")

# 5. Self-consistency style
run("""
Why does ice float on water?

Let's think step by step:
""")

# 6. Instruction prompting
run("""
Summarize the following in 3 lines:
Artificial Intelligence enables machines to learn from data and make decisions.
""")

# 7. Role prompting
run("""
You are a polite science teacher.
Explain how the Internet works.
""")

# 8. Delimiter prompting
run("""
Summarize this text: <text>Machines learn patterns from data in ML.</text>
""")

# 9. Contrastive prompt
run("""
Bad summary: It's about machines.
Good summary: Give a precise, clear summary.

Text: Deep learning uses neural networks with layers to learn patterns from data.

Summary:
""")
```

Compare retrieval-based vs generative chatbot responses.

```python
!pip install -q transformers scikit-learn

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

corpus = [
    "Hello, how can I help you?",
    "AI stands for Artificial Intelligence.",
    "Machine learning is a subset of AI.",
    "The weather is nice today.",
    "Chatbots can be retrieval-based or generative."
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

def retrieval_chatbot(query):
    q_vec = vectorizer.transform([query])
    sims = cosine_similarity(q_vec, X).flatten()
    return corpus[sims.argmax()]

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

def generative_chatbot(query):
    inputs = tokenizer(query, return_tensors="pt")
    output = model.generate(**inputs, max_new_tokens=50)
    return tokenizer.decode(output[0], skip_special_tokens=True)

query = "What is AI?"
```

```python
print("User:", query)
print("\nRetrieval-based Response:\n", retrieval_chatbot(query))
print("\nGenerative Response:\n", generative_chatbot(query))
```
