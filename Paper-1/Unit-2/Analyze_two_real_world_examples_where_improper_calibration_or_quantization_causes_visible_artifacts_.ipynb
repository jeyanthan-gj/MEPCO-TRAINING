Color Banding Due to Low Bit-Depth Quantization

```python
import numpy as np
import matplotlib.pyplot as plt

# Create smooth gradient
x = np.linspace(0, 1, 1024)
gradient = np.tile(x, (256, 1))

# Quantize to low bit-depth (4 bits = 16 levels)
levels = 16
quantized = np.floor(gradient * levels) / levels

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.title("Original Smooth Gradient")
plt.imshow(gradient, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Quantized (Banding Appears)")
plt.imshow(quantized, cmap='gray')
plt.axis('off')

plt.show()
```

Incorrect White Balance Calibration Causing Color Cast

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/istockphoto-480931380-612x612.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Simulate poor white balance (too blue)
wb_wrong = img.astype(float)
wb_wrong[:,:,2] *= 1.3  # increase blue
wb_wrong = wb_wrong.clip(0,255).astype('uint8')

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(img)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Improper White Balance (Color Cast)")
plt.imshow(wb_wrong)
plt.axis('off')

plt.show()
```
