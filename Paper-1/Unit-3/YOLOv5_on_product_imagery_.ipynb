```python
from ultralytics import YOLO

# Load pretrained YOLOv5s model
model = YOLO("yolov5s.pt")  # small model, fast training

# Train using the uploaded YAML
results = model.train(
    data="SKU-110K.yaml",  # your uploaded YAML
    epochs=1,
    imgsz=640,
    batch=16,
    name="SKU110K_v5"
)
```

```python
from ultralytics import YOLO
from IPython.display import display
import glob, random
import matplotlib.pyplot as plt

# Load your trained YOLOv5su model
model = YOLO("/content/yolov5/yolov5/datasets/runs/detect/SKU110K_v52/weights/best.pt")

# List all test images
test_images = glob.glob("/content/yolov5/yolov5/datasets/SKU-110K/images/test_*.jpg")

# Pick 5 random images
sample_images = random.sample(test_images, 5)

# Run inference and display predictions inline
for img_path in sample_images:
    results = model.predict(source=img_path, conf=0.25)  # Run inference
    # Plot the image with bounding boxes
    fig = results[0].plot()  # returns numpy array
    plt.figure(figsize=(12,8))
    plt.imshow(fig)
    plt.axis('off')
    plt.show()
```
